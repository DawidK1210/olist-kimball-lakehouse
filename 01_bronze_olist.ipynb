```python
from pyspark.sql import functions as F

raw_path = "/Volumes/workspace/olist_bronze/raw/"
files = [f for f in dbutils.fs.ls(raw_path) if f.name.endswith(".csv")]

for f in files:
    name = f.name.replace(".csv", "").replace("olist_", "").replace("_dataset", "")
    table = f"workspace.olist_bronze.bronze_{name}"

    df = (spark.read
          .option("header", True)
          .option("multiline", True)
          .option("escape", '"')
          .csv(f.path)
          .select("*",
                  F.col("_metadata.file_path").alias("_source_file"),
                  F.current_timestamp().alias("_ingested_at")))
    
    df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(table)
    print(table, spark.table(table).count())
```
